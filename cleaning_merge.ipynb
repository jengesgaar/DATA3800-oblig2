```python
import pandas as pd

# Read the file into a DataFrame
crop_df = pd.read_csv(
    "datasets/04610.csv",
    sep=";",
    skiprows=1,
    encoding="cp1252"
)

print("Rows: ", crop_df.shape[0])
print("Columns: ", crop_df.shape[1])
print("Column Names:")
print(crop_df.columns)

display(crop_df.head())
```

```python
# Inspect the columns and data types
print("First 5 rows: ")
print(crop_df.iloc[:5])

print("Last 5 rows: ")
print(crop_df.iloc[-5:])

crop_df.info()
```

```python
# Count special values in crop data
crop_values = crop_df.drop(columns="region").stack()

print("Special values: ")
print(crop_values.value_counts().reindex([".", "..", ":", "-", "0"], fill_value=0))
```

```python
# Find values that are neither numbers nor special values
numeric_values = pd.to_numeric(crop_values, errors="coerce")

other_values = crop_values[
    numeric_values.isna()
    & ~crop_values.isin([".", "..", ":", "-"])
]

print("Other values: ")
print(other_values.value_counts())
```

```python
# Replace special values with missing values in a new DataFrame
crop_clean_df = crop_df.replace([".", "..", ":", "-"], pd.NA)

print("Missing values: ")
print(crop_clean_df.drop(columns="region").isna().sum().sum())

display(crop_clean_df.head())
```

```python
# Convert crop yield columns from text to numbers
crop_clean_df[crop_clean_df.columns[1:]] = crop_clean_df[
    crop_clean_df.columns[1:]
].apply(pd.to_numeric, errors="coerce")

print("Data types: ")
print(crop_clean_df.dtypes.value_counts())

print("\nMissing values: ")
print(crop_clean_df.drop(columns="region").isna().sum().sum())
```

```python
# Reshape the crop data from wide to long format for increased readability and easier analysis
crop_long_df = crop_clean_df.melt(
    id_vars="region",
    var_name="year_crop",
    value_name="yield_kg_per_decare"
)

print("Rows: ", crop_long_df.shape[0])
print("Columns: ", crop_long_df.shape[1])

display(crop_long_df.head())
display(crop_long_df.tail())
```

```python
# Split the year and crop type into separate columns
crop_long_df[["year", "crop"]] = crop_long_df["year_crop"].str.split(
    " ", n=1, expand=True
)

# Convert year from text to an integer
crop_long_df["year"] = crop_long_df["year"].astype(int)

display(crop_long_df[["year_crop", "year", "crop"]].head())
display(crop_long_df[["year_crop", "year", "crop"]].tail())
```

```python
# Check the crop types and year range
print("Crop types: ")
print(crop_long_df["crop"].unique())

print("\nFirst year: ", crop_long_df["year"].min())
print("Last year: ", crop_long_df["year"].max())

print("\nMissing values in yield: ")
print(crop_long_df["yield_kg_per_decare"].isna().sum())
```

```python
# Keep the final columns in a clear order
crop_final_df = crop_long_df[
    ["region", "year", "crop", "yield_kg_per_decare"]
].copy()

print("Rows: ", crop_final_df.shape[0])
print("Missing yield values: ", crop_final_df["yield_kg_per_decare"].isna().sum())

display(crop_final_df.head())
```

```python
# Split region code? and region name for future merging
crop_final_df[["region_code", "region_name"]] = crop_final_df["region"].str.split(
    " ", n=1, expand=True
)

display(
    crop_final_df[
        ["region", "region_code", "region_name"]
    ].drop_duplicates()
)
```

```python
# Select and order the columns for the cleaned dataset
crop_final_df = crop_final_df[
    ["region_code", "region_name", "year", "crop", "yield_kg_per_decare"]
].copy()

print("Rows: ", crop_final_df.shape[0])
print("Columns: ", crop_final_df.shape[1])
print("Duplicate region-year-crop combinations: ",
      crop_final_df.duplicated(["region_code", "year", "crop"]).sum())

display(crop_final_df.head())
```

```python
# Save the cleaned dataset as a new file
crop_final_df.to_csv(
    "datasets/04610_cleaned.csv",
    index=False,
    encoding="utf-8-sig",
)
```
